```python
import os

# Define the directory containing the CSV files, ensure it is a string
directory = os.fsencode('/content/65degreeC_1')

# Get a list of all files in the directory that end with .csv
csv_files = [f for f in os.listdir(directory) if f.endswith(b".csv")]  # Use b".csv" for bytes comparison

# Sort the files based on the numerical value in the file name
csv_files.sort(key=lambda x: int(os.path.splitext(x)[0].decode("utf-8")))  # Decode the filename for sorting

# Rename the files sequentially
for i, filename in enumerate(csv_files, start=1):
    new_name = f"{i}.csv".encode()  # Encode new_name to bytes
    old_path = os.path.join(directory, filename)
    new_path = os.path.join(directory, new_name)
    os.rename(old_path, new_path)

print("Files have been renamed successfully.")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Assuming 'combined_df_merged' is your DataFrame
# Filter the DataFrame for the specific file number '13_80_1'
filtered_df = combined_df_merged[combined_df_merged['File_no'] == '13_80_1']

# Further filter the DataFrame based on the time range
filtered_df = filtered_df[(filtered_df['Time'] >= 14250) & (filtered_df['Time'] <= 32500)]

# Identify the first peak point (local maximum)
leser_d_values = filtered_df['Leser_D'].values
time_values = filtered_df['Time'].values

# Calculate the differences to identify the first decline after a peak
diffs = np.diff(leser_d_values)

# Find the first index where the difference is negative (indicating a peak)
peak_index = np.where(diffs < 0)[0][0] + 1  # +1 because np.diff reduces the length by 1

# Find the second significant change after the first peak (where the slope changes again)
# We're looking for the next point where the slope is near zero after a substantial change
second_change_index = np.where(np.abs(diffs[peak_index:]) > 0)[0][0] + peak_index

# Extract the peak time and position
peak_time = time_values[peak_index]
peak_value = leser_d_values[peak_index]

# Extract the second point time and position
second_change_time = time_values[second_change_index]
second_change_value = leser_d_values[second_change_index]

# Plotting the filtered DataFrame
plt.plot(filtered_df['Time'], filtered_df['Leser_D'], label='Injection Position')

# Mark the first peak on the plot
plt.plot(peak_time, peak_value, 'ro', label='First Peak')
plt.axvline(peak_time, color='r', linestyle='--', label=f'First Peak at {peak_time}')

# Mark the second significant change on the plot
plt.plot(second_change_time, second_change_value, 'bo', label='Second Change')
plt.axvline(second_change_time, color='b', linestyle='--', label=f'Second Change at {second_change_time}')

# Adding labels and title for clarity
plt.xlabel('Time')
plt.ylabel('Injection Position')
plt.title('Time vs Injection Position for File_no 13_80_1')
plt.legend()

# Displaying the plot
plt.show()

print(f"The first peak occurs at time {peak_time} with a position of {peak_value}.")
print(f"The second significant change occurs at time {second_change_time} with a position of {second_change_value}.")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, WhiteKernel, ConstantKernel as C
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, r2_score

# Assuming 'agg_df' is loaded from your dataset
X = agg_df.drop(columns=['Microns_mean', 'File_no_'])
y = agg_df['Microns_mean']

# Split the data
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Define the kernel and Gaussian Process Regressor

#Kernel: C(1.5, (1e-9, 1e9)) * Periodic(length_scale=) + C(1.5, (1e-9, 1e9)) * Matern(length_scale=23.063401978990722, nu=nu) + WhiteKernel(5.24)
kernel = C(1.5, (1e-13, 1e13)) * Matern(length_scale=23.063, nu=1.5) + WhiteKernel(noise_level=5.24)

model = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=100, random_state=42)

# Train the model
model.fit(X_train_scaled, y_train)

# Make predictions
y_pred, y_std = model.predict(X_test_scaled, return_std=True)

# Estimate aleatoric uncertainty from the model's learned noise level
aleatoric_uncertainty = np.sqrt(model.kernel_.k2.noise_level)  # Extract the learned noise level

# Calculate total uncertainty (epistemic + aleatoric)
total_uncertainty = np.sqrt(y_std**2 + aleatoric_uncertainty**2)

# Evaluate the model
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f'Mean Absolute Error: {mae}')
print(f'R-squared: {r2}')

# Plot predictions with uncertainties on actual test data
plt.figure(figsize=(15, 6))

# Plot predictive mean and total uncertainty
#plt.errorbar(range(len(y_test)), y_pred, yerr=total_uncertainty, fmt='o', label='Predictive mean ± Total uncertainty', capsize=5, capthick=2)
plt.fill_between(range(len(y_test)), y_pred - y_std - total_uncertainty, y_pred + y_std + total_uncertainty, alpha=0.2, label='Total uncertainty', color='C1')

# Fill area for epistemic uncertainty (GP variance)
plt.fill_between(range(len(y_test)), y_pred - y_std, y_pred + y_std, alpha=0.2, label='Epistemic uncertainty', color='blue')

# Plot actual test data
plt.scatter(range(len(y_test)), y_test, color='red', label='Actual test data', marker='x')

# Labels and title
plt.xlabel('Test data index')
plt.ylabel('Microns_mean')
plt.title('Prediction vs Actual with Uncertainty on Test Data')
plt.legend()

# Show the plot
plt.show()
```
